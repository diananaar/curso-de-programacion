```python
# Datos asignados para tu reto
portafolio = {
    "MSFT": {"cantidad": 10, "precio_compra": 109},
    "V": {"cantidad": 8, "precio_compra": 153},
    "META": {"cantidad": 12, "precio_compra": 127},
    "NVDA": {"cantidad": 6, "precio_compra": 492}
}

# Lista de activos
activos = list(portafolio.keys())

print("Activos:", activos)

print("informacion de MSFT",portafolio ["MSFT"])


```

```python
precios_actuales = {
    "MSFT": 117,
    "V": 167,
    "META": 109,
    "NVDA": 546
}

print(precios_actuales)
```

```python
cantidad = portafolio["MSFT"]["cantidad"]
precio_compra = portafolio["MSFT"]["precio_compra"]
precio_actual = precios_actuales["MSFT"]

valor_invertido = cantidad * precio_compra
valor_actual = cantidad * precio_actual

print("Valor invertido:", valor_invertido)
print("Valor actual:", valor_actual)
```

```python
for activo in activos:
    cantidad = portafolio[activo]["cantidad"]
    precio_compra = portafolio[activo]["precio_compra"]
    precio_actual = precios_actuales[activo]

    valor_invertido = cantidad * precio_compra
    valor_actual = cantidad * precio_actual

    print(activo, "| Invertido:", valor_invertido, "| Actual:", valor_actual)
```

```python
ganancia = valor_actual - valor_invertido
rendimiento = (ganancia / valor_invertido) * 100

print("Ganancia / pérdida:", ganancia)
print("Rendimiento:", round(rendimiento, 2), "%")
```

```python
if rendimiento >= 10:
    recomendacion = "MANTENER"
elif rendimiento >= 0:
    recomendacion = "REVISAR"
else:
    recomendacion = "ANALIZAR VENTA"

print("Recomendación:", recomendacion)
```

```python
def analizar_activo(cantidad, precio_compra, precio_actual):
    valor_invertido = cantidad * precio_compra
    valor_actual = cantidad * precio_actual
    ganancia = valor_actual - valor_invertido
    rendimiento = (ganancia / valor_invertido) * 100

    return valor_invertido, valor_actual, ganancia, rendimiento


# Prueba de la función
resultado = analizar_activo(10, 180, 195)
print(resultado)
```

```python
resultados = []

for activo in activos:
    cantidad = portafolio[activo]["cantidad"]
    precio_compra = portafolio[activo]["precio_compra"]
    precio_actual = precios_actuales[activo]

    valor_invertido, valor_actual, ganancia, rendimiento = analizar_activo(
        cantidad, precio_compra, precio_actual
    )

    if rendimiento >= 10:
        recomendacion = "MANTENER"
    elif rendimiento >= 0:
        recomendacion = "REVISAR"
    else:
        recomendacion = "ANALIZAR VENTA"

    resultados.append({
        "activo": activo,
        "valor_invertido": valor_invertido,
        "valor_actual": valor_actual,
        "ganancia": ganancia,
        "rendimiento": rendimiento,
        "recomendacion": recomendacion
    })

resultados
```

```python
total_invertido = 0
total_actual = 0

print("=" * 65)
print("ANÁLISIS DEL PORTAFOLIO")
print("=" * 65)

for resultado in resultados:
    total_invertido += resultado["valor_invertido"]
    total_actual += resultado["valor_actual"]

    print(
        f"{resultado['activo']:6} | "
        f"Actual: ${resultado['valor_actual']:,.0f} | "
        f"Rendimiento: {resultado['rendimiento']:.2f}% | "
        f"{resultado['recomendacion']}"
    )

ganancia_total = total_actual - total_invertido
rendimiento_total = (ganancia_total / total_invertido) * 100

print("=" * 65)
print(f"Total invertido: ${total_invertido:,.0f}")
print(f"Valor actual:    ${total_actual:,.0f}")
print(f"Ganancia/pérdida: ${ganancia_total:,.0f}")
print(f"Rendimiento:     {rendimiento_total:.2f}%")
```

```python
mejor_activo = max(resultados, key=lambda x: x["rendimiento"])

print("Activo con mejor rendimiento:", mejor_activo["activo"])
print("Rendimiento:", round(mejor_activo["rendimiento"], 2), "%")
print("Recomendación:", mejor_activo["recomendacion"])
```
